```python
def run_market_sizing():
    # --- Top-Down Model Inputs ---
    population = 3_500_000  # e.g., professionals in UAE
    internet_pen = 0.98     # 98%
    target_seg = 0.30       # 30% are decision makers
    conv_rate = 0.05        # 5% conversion
    annual_arpu = 1200      # $1,200 per year

    # Calculation
    tam_users = population * internet_pen
    sam_users = tam_users * target_seg
    som_users = sam_users * conv_rate
    revenue_estimate = som_users * annual_arpu

    # --- Output ---
    print("="*45)
    print("  STRATEGIC MARKET SIZING MODEL (UAE)")
    print("="*45)
    print(f"{'Total Addressable Users:':<30} {tam_users:,.0f}")
    print(f"{'Serviceable Users:':<30} {sam_users:,.0f}")
    print(f"{'Obtainable Customers (SOM):':<30} {som_users:,.0f}")
    print("-" * 45)
    print(f"{'ESTIMATED ANNUAL REVENUE:':<30} ${revenue_estimate:,.2f}")
    print("="*45)

    # Sensitivity Table
    print("\nREVENUE SENSITIVITY (Conv. Rate vs ARPU)")
    print(f"{'Conv %':<10} | {'$1,000 ARPU':<12} | {'$1,200 ARPU':<12} | {'$1,400 ARPU':<12}")
    for rate in [0.03, 0.05, 0.07]:
        row = f"{rate:<10.1%} | "
        for price in [1000, 1200, 1400]:
            val = (sam_users * rate) * price
            row += f"${val/1e6:>10.1f}M | "
        print(row)

run_market_sizing()
```

```python
def triangulated_market_sizing():
    # --- Top-Down Model ---
    pop = 3_500_000
    target_penetration = 0.40
    conversion = 0.05
    price_point = 1500

    top_down_som = (pop * target_penetration * conversion) * price_point

    # --- Bottom-Up Model ---
    sales_reps = 12
    leads_per_rep_monthly = 50
    close_rate = 0.10
    avg_deal_size = 18000

    bottom_up_som = (sales_reps * leads_per_rep_monthly * 12 * close_rate) * avg_deal_size

    # --- Analysis & Triangulation ---
    gap = abs(top_down_som - bottom_up_som)
    avg_estimate = (top_down_som + bottom_up_som) / 2
    confidence = "HIGH" if (gap / avg_estimate) < 0.2 else "LOW (Review Assumptions)"

    print("█" * 60)
    print("   MARKET SIZING TRIANGULATION: B2B SAAS (UAE)")
    print("█" * 60)

    print(f"\n[MODEL 1] Top-Down (Population Based):    ${top_down_som/1e6:>6.2f}M")
    print(f"[MODEL 2] Bottom-Up (Sales Capacity):     ${bottom_up_som/1e6:>6.2f}M")
    print("-" * 60)
    print(f"TRIANGULATED ESTIMATE (AVG):              ${avg_estimate/1e6:>6.2f}M")
    print(f"MODEL VARIANCE (THE GAP):                 ${gap/1e6:>6.2f}M")
    print(f"DATA CONFIDENCE LEVEL:                    {confidence}")

    print("\n[STRATEGIC RECOMMENDATION]")
    if confidence == "HIGH":
        print(" • The models align. Proceed with GTM strategy based on $100M+ opportunity.")
    else:
        print(" • Significant Gap detected. Bottom-up sales capacity outpaces top-down demand.")
        print(" • Action: Audit 'Conversion Rate' (5%) vs 'Close Rate' (10%) for consistency.")

triangulated_market_sizing()
```
